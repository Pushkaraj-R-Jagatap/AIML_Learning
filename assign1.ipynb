```python
import json

#laod the file
def load_data(filename):
    with open(filename,"r")as f:
        data=json.load(f)
    return data
    
```

```python
data=load_data("store_data.json")
```

```python
print(type(data))
print(data)
```

```python
#clena and structue the data

def clean_data(data):
    
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    cleaned_data=[]
    unique_users=set()
    
    for user in data:
        raw_rating=str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
        user["rating"]=raw_rating


        #handle missing value
        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None
            
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
```

```python
data=clean_data(data)
```

```python
#to get the meaning ful insights
def get_insights(data):
    tat_rating=0
    for user in data:
        tat_rating +=float(user['rating'])
        # print("the tat_rating=",tat_rating)
    print(f"The  avg rating is={tat_rating/len(data)}")


    #calculate the poor rating
    poor_rating=0
    for user in data:
        if(float(user["rating"])<3):
            poor_rating+=1

    print(f"The user count given poor rating={poor_rating/len(data)*100}%")
```

```python
get_insights(data)
```
