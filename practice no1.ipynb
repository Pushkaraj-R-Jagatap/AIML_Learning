```python
import json
```

```python
def load_data(filename):
    with open(filename,"r")as f:
        data=json.load(f)
    return data
    
```

```python
data=load_data("student_data.json")
```

```python
def clean_data(data):
    text_to_number={"one":1,"two":2,"three":3,"four":4,"five":5,"ninety":90}
    cleaned_data=[]
    unique_user=set()
    for user in data:
         
         raw_marks=str(user["marks"]).strip().lower()
         if(raw_marks in text_to_number):
             user["marks"]=text_to_number[raw_marks]
         else:
             user["marks"]=int(raw_marks)

         if user.get("age") is None:
             user["age"]="None"
         if(user["name"]in unique_user):
             continue
         unique_user.add(user["name"])
         cleaned_data.append(user)
         
    return cleaned_data

    
```

```python
clean_data=clean_data(data)

clean_data
```

```python
def insights(data):
    total_marks=0
    passing_count=0
    poor_perf=0
    missing_age=0
    highest_score=0
    course_count={}
    for user in data:
        total_marks+=user["marks"]
    print(f"average marks of students={total_marks/len(data)}")

    for user in data:
        if(user["marks"]>=40):
            passing_count+=1
        else:
            poor_perf+=1

        if(user.get("age")==None):
            missing_age+=1
            
    

        if(user["marks"]>highest_score) :
            highest_score=user["marks"]
            highest_score_name=user["name"]
        course=user["course"]
        if course in course_count:
            course_count[course]+=1
        else:
            course_count[course]=1
            
        
    print(f"Total number of students passed={passing_count}")
    print(f"Total number of students have poor performance={poor_perf}")
    print(f"The highest score is{highest_score} and his name is{highest_score_name}")
    print(f"The course wise student count={course_count}")
    
        

        
        

        
        
```

```python
insights(clean_data)
```
