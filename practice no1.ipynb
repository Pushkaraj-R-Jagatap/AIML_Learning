```python
import json
```

```python
def load_data(filename):
    with open(filename,"r")as f:
        data=json.load(f)
    return data
    
```

```python
data=load_data("student_data.json")
```

```python
print(data,type(data))
print(len(data))
```

```python
for user in data:
    user["name"]=user["name"].strip().lower()
```

```python
def clean_data(data):
    text_to_number={"one":1,"two":2,"three":3,"four":4,"five":5,"ninety":90}

    for user in data:
         
         raw_marks=str(user["marks"]).strip().lower()
         if(raw_marks in text_to_number):
             user["marks"]=text_to_number[raw_marks]
         else:
             user["marks"]=int(raw_marks)
         
    return data
```

```python
clean_data(data)
```
